## Import Modules

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

## Load in Files

```python
df = pd.read_csv('../Data/Results/NoPCA_vs_PCA.csv')
AEI = pd.read_csv('../Data/Results/NoAEI_vs_AEI.csv')
model = pd.read_csv('../Data/Results/Models_PCA_vs_none.csv')
sp = pd.read_csv('../Data/Results/Specific_disorder_stats.csv')
```

## Plot AUCs for No PCA vs PCA

#### This creates a line plot for the average AUC values of logistic regression for each main disorder, for both the models trained on PCA and without. The standard deviations of the AUCs are the error bars.

```python
disorders = df.columns[1:]  # skip the first column
x = np.arange(len(disorders))

# Extract rows
without_pca = df.iloc[0, 1:].astype(float)
with_pca = df.iloc[1, 1:].astype(float)
std_without = df.iloc[2, 1:].astype(float)
std_with = df.iloc[3, 1:].astype(float)

# Plot with error bars
plt.figure(figsize=(10, 6))
plt.errorbar(x, without_pca, yerr=std_without, label='Without PCA', 
             marker='o', capsize=5, linestyle='-', color='lightsteelblue')
plt.errorbar(x, with_pca, yerr=std_with, label='With PCA', 
             marker='o', capsize=5, linestyle='-', color='slategray')

# Final touches
plt.xticks(x, disorders, rotation=45, ha='right')
plt.ylabel("AUC Value")
plt.xlabel("Disorder")
plt.title("Logistic Regression Average AUC Values for Different Main Disorders")
plt.ylim(0, 1)
plt.legend()
plt.tight_layout()
plt.show()
```
