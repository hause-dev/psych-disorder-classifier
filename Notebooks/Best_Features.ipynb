## Import and Load Desired Files

```python
import pandas as pd

# Load your CSV file
df = pd.read_csv('/home/novo/psych-disorder-classifier/Data/Results/Feature_Combination_Results.csv')
sp = pd.read_csv('/home/novo/psych-disorder-classifier/Data/Results/specific_disorder_data.csv')
```

## Finding Averages and Standard Deviation for Each Algorithm

```python
df_numeric = sp.iloc[:, 2:]

every_other_col = df_numeric.iloc[:, ::2]

# Split: even rows = without PCA, odd rows = with PCA
without_pca = every_other_col.iloc[::2]
with_pca = every_other_col.iloc[1::2]

# Calculate stats
avg_no_pca = without_pca.mean()
avg_with_pca = with_pca.mean()
std_no_pca = without_pca.std()
std_with_pca = with_pca.std()

results_df = pd.DataFrame({
    'Without PCA': avg_no_pca,
    'With PCA': avg_with_pca,
    'SD No PCA': std_no_pca,
    'SD PCA': std_with_pca
})

# Transpose so algorithm names become column headers
results_sp = results_df.T

# Save to CSV
results_sp.to_csv("/home/novo/psych-disorder-classifier/Data/Results/Specific_disorder_stats.csv")
```

## Finding Best Features

```python
row_labels = df.iloc[:, 0]
data = df.iloc[:, 2::2]  # Skip first 2 columns and select every other column

disorders = row_labels.str.split('_').str[0].unique()
all_results = []

for disorder in disorders:
    disorder_mask = row_labels.str.contains(f'^{disorder}_', case=False, regex=True)
    disorder_rows = df[disorder_mask]
    disorder_labels = disorder_rows.iloc[:, 0]
    disorder_data = disorder_rows.iloc[:, 2::2]

    for col in disorder_data.columns:
        max_val = disorder_data[col].max()
        matching_rows = disorder_data[disorder_data[col] == max_val]
        matching_labels = disorder_labels.loc[matching_rows.index].tolist()

        result = pd.DataFrame({
            'Disorder': [disorder] * len(matching_labels),
            'Feature': [col] * len(matching_labels),
            'MaxValue': [max_val] * len(matching_labels),
            'RowLabel': matching_labels
        })

        all_results.append(result)

final_df = pd.concat(all_results, ignore_index=True)
final_df.to_csv('/home/novo/psych-disorder-classifier/Data/Results/Best_Feature_Combinations.csv', index=False)
```
